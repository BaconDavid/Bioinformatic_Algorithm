## Code Challenge: Solve the String Composition Problem.

- Input: An integer k and a string Text.
- Output: Compositionk(Text) (the k-mers can be provided in any order).

Sample Input 1:

5
CAATCCAAC
Sample Output 1:

CAATC AATCC ATCCA TCCAA CCAAC

# Solution 3.2

```python
import sys
from typing import List, Dict, Iterable

# Please do not remove package declarations because these are used by the autograder.

# Insert your kmer_composition function here, along with any subroutines you need
def kmer_composition(text: str, k: int) -> Iterable[str]:
    """Forms the k-mer composition of a string."""
    text_length = len(text)
    num_kmers = text_length - k + 1
    kmers = [text[i:i+k] for i in range(num_kmers)]
    return kmers
test_text = 'GGGGGGTGGG'
test_k = 3
kmer_composition(test_text, test_k)
```

```python
import sys
from typing import List, Dict, Iterable

# Please do not remove package declarations because these are used by the autograder.

# Insert your genome_path function here, along with any subroutines you need
def genome_path(path: List[str]) -> str:
    """Forms the genome path formed by a collection of patterns."""
    for i in range(len(path)):
        if i == 0:
            text = path[i]
        else:
            text += path[i][-1]
    return text

paths = ['TT', 'TG', 'GT', 'TT']
genome_path(paths)


```

## Code Challenge: Solve the Overlap Graph Problem (restated below).

Input: A collection Patterns of k-mers.
Output: The overlap graph Overlap(Patterns), in the form of an adjacency list. (You may return the nodes and their edges in any order.)

```python
import sys
from typing import List, Dict, Iterable
import numpy as np
# Please do not remove package declarations because these are used by the autograder.

# Insert your overlap_graph function here, along with any subroutines you need
def overlap_graph(patterns: List[str]) -> Dict[str, List[str]]:
    """Forms the overlap graph of a collection of patterns."""
    
    # Initialize the overlap graph dictionary with empty lists
    graph = {}
    
    # Process each pattern
    for pattern in patterns:
        if pattern not in graph:
            graph[pattern] = []
        
        # Get the suffix of the current pattern (all but the first character)
        suffix = pattern[1:]
        
        # Check all patterns for a prefix match with our suffix
        for potential_target in patterns:
            # Get the prefix of the potential target (all but the last character)
            prefix = potential_target[:-1]
            
            # If suffix of pattern matches prefix of potential_target, add an edge
            if (suffix == prefix) and (potential_target not in graph[pattern]):
                graph[pattern].append(potential_target)
    
    # Filter out patterns with no outgoing edges
    return {k: v for k, v in graph.items() if v}

input_patterns = ['AAG', 'AGA', 'ATT', 'CTA', 'CTC', 'GAT', 'TAC', 'TCT', 'TCT', 'TTC']
overlap_graph(input_patterns)





```
